```python
data = pd.read_csv("bank.csv",sep=";")
data
```

```python
ie =preprocessing.LabelEncoder()
job = ie.fit_transform(list(data["job"]))
marital = ie.fit_transform(list(data["marital"]))
education = ie.fit_transform(list(data["education"]))
default = ie.fit_transform(list(data["default"]))
housing = ie.fit_transform(list(data["housing"]))
loan = ie.fit_transform(list(data["job"]))
contact = ie.fit_transform(list(data["contact"]))
month = ie.fit_transform(list(data["month"]))
poutcome = ie.fit_transform(list(data["poutcome"]))
yes = ie.fit_transform(list(data["y"]))
```
